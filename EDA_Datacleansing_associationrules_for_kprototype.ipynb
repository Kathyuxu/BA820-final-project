## **Data Cleaning**

```python
merged_data_amentities = merged_data[["amenities","cluster_label"]]
merged_data_amentities.isnull().sum()
merged_data_amentities.dropna(subset=['cluster_label'], inplace = True)
merged_data_amentities.info()
merged_data_amentities.duplicated().sum()
merged_data_amentities.drop_duplicates(inplace = True)
merged_data_amentities.info()
```

```python
import ast

# Convert string representation of lists into actual lists
merged_data_amentities["amenities"] = merged_data_amentities["amenities"].apply(lambda x: ast.literal_eval(x) if isinstance(x, str) else x)

# Remove items containing "%\\%" from each list
merged_data_amentities["amenities"] = merged_data_amentities["amenities"].apply(lambda x: [item for item in x if "%\\%" not in str(item)] if isinstance(x, list) else x)
```

```python
# divide the group into 3 clusters
cluster_1 = merged_data_amentities[merged_data_amentities['cluster_label'] == 0.0]
cluster_2 = merged_data_amentities[merged_data_amentities['cluster_label'] == 1.0]
cluster_3 = merged_data_amentities[merged_data_amentities['cluster_label'] == 2.0]
```

## **Association Rules**

```python
cluster_1_test = cluster_1["amenities"].tolist()
cluster_2_test = cluster_2["amenities"].tolist()
cluster_3_test = cluster_3["amenities"].tolist()
```

```python
from mlxtend.frequent_patterns import apriori
from mlxtend.frequent_patterns import association_rules
from mlxtend.preprocessing import TransactionEncoder
from collections import Counter

def flatten_list(lst, min_n, max_n):
    flat_list = []
    for sublist in lst:
        if isinstance(sublist, list):
            flat_list += sublist
        else:
            flat_list.append(sublist)

    counter = Counter(flat_list)

    max_n = min(max_n, len(counter))
    min_n = min(min_n, max_n)

    top_values = [item[0] for item in counter.most_common(max_n)[min_n-1:max_n]]

    filtered_data = [[item for item in amenities if item in top_values] for amenities in lst]

    te = TransactionEncoder()
    transactions = te.fit_transform(filtered_data)

    df_encoded = pd.DataFrame(transactions, columns=te.columns_)
    return df_encoded
```

```python
flatten_list_1 = flatten_list(cluster_1_test, 12, 27)
flatten_list_1
```

```python
frequent_itemsets_1 = apriori(flatten_list_1, min_support=0.5, use_colnames=True)
frequent_itemsets_1.sort_values(by="support", ascending=False)
```

```python
rules_1 = association_rules(frequent_itemsets_1, metric = "support", min_threshold=0.05)
rules_1
```
